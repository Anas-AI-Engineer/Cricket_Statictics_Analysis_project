```python
import numpy as np

print("-----Cricket statictics analyzer-----")

players =([
    "BabarAzam",
    "Rizwan",
    "FakherZaman",
    "SaimAyub",
    "ShadabKhan",
    "MuhammadAmir",
    "ShaheenAfridi",
    "HarisRauf"
])

matches = np.array([1, 2, 3, 4, 5, 6, 7, 8,])
runs = np.array([1000, 1500, 2000, 2500, 1500, 700, 200, 250])
balls_faced = np.array([100, 150, 200, 250, 150, 70, 10, 20,])
dismissals = np.array([30, 50, 40, 75, 32, 70, 3, 10,])
four = np.array([20, 25, 30, 35, 6, 3, 1, 1,])
six = np.array([10, 20, 30, 40, 1, 1, 1, 1,])

wickets = np.array([0, 0, 1, 0, 5, 4, 6, 3,])
overs = np.array([1, 2, 1, 2, 3, 2, 4, 5,])
runs_conceded = np.array([15, 20, 10, 20, 60, 60, 40, 46,])

batting_average = np.divide(
    runs,dismissals,
    out=np.zeros(len(players), dtype=float),
    where = dismissals !=0
)

strike_rate = np.divide(
    runs *100, balls_faced,
    out =np.zeros(len(players), dtype=float),
    where=balls_faced !=0
)
boundry_runs = (four*4) + (six*6)

economy = np.divide(
    runs_conceded, overs,
    out= np.zeros(len(players), dtype=float),
    where=overs !=0
)

bowling_average = np.divide(
    runs_conceded, wickets,
    out = np.zeros(len(players), dtype=float),
    where = wickets !=0
)

for i in range(len(players)):
    print(f"Players: {palyers[i]}")
    print(f"Matches: {matches[i]}")
    print(f"Runs: {runs[i]}")
    print(f"Strike_rate: {strike_rate[i]:.2f}")
    print(f"Batting_Average: {batting_average[i]:.2f}")
    print(f"Boundry_runs: {boundry_runs[i]:.2f}")
    print(f"Wickets: {wickets[i]}")
    print(f"Bowling Economy: {economy[i]:.2f}")
    print(f"Bowling Average: {bowling_average[i]:.2f}")

print("======Team Batting Summary=====")
print("Total runs:", np.sum(runs))
print("Total Fours:", np.sum(four))
print("Total Sixes:", np.sum(six))
print("Total Boundry Runs:", np.sum(boundry_runs))

print("=====Find Top scocer====")
top_score_index = np.argmax(runs)
print("Top Scocer:", players[top_score_index])
print("Top Runs:", runs[top_score_index])

print("====Top Wicket Taker====")
top_wicket_taker = np.argmax(wickets)
print("Top Wickets:", players[top_wicket_taker])
print("Wickets:", wickets[top_wicket_taker])


print("====Find MOst Economical Bowler Minimium 5 over====")

eligibal_bowlers = overs >= 5

if np.any(eligibal_bowlers):
    eligibal_indices = np.where(eligibal_bowlers)[0]
    best_economy_index= eligibal_indices[
        np.argmin(economy[eligibal_bowlers])
    ]

    print("Most Economical Bowlers:", players[best_economy_index])
    print("Economy:", round(economy[best_economy_index],2))

print("====Find the players how score 100 more runs====")
for i in np.where(runs >= 100)[0]:
    print(players[i],  "-", runs[i], "runs")



    

                  
```
